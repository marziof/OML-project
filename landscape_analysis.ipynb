```python
seed_kept = [] 
for seed in range(50):

    rng = np.random.default_rng(seed)
    x0 = rng.uniform(-5, 5, size=2)

    sgd_final, sgd_traj = plain_gd(
        objectives.rastrigin_gd,
        x_init=x0.copy(),
        eta=0.001,
        num_epochs=2000
    )

    center_final, center_traj, _ = seasgd(
        objectives.rastrigin_gd,
        objectives.rastrigin_fn,
        theta=x0.copy(),
        **seasgd_kwargs_rastri
    )
    sgd_loss = objectives.rastrigin_fn(sgd_final)
    seasgd_loss = objectives.rastrigin_fn(center_final)
    if seasgd_loss < sgd_loss:
        seed_kept.append([seed,sgd_traj,center_traj, sgd_loss,seasgd_loss])
for i in range(3):
    j = random.randint(0,len(seed_kept)-1)
    print(
            f"seed={seed_kept[j][0]}, "
            f"SGD={seed_kept[j][3]:.2f}, "
            f"SEASGD={seed_kept[j][4]:.2f}"
        )
    plot_master_vs_sgd(
        objectives.rastrigin_fn,
        seed_kept[j][1],
        seed_kept[j][2],
        title=f"Seed {seed_kept[j][0]}"
    )
```
